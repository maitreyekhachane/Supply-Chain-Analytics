### Shipping Time Validation

The actual and scheduled shipping-time columns were checked for missing and invalid values.

No missing or negative values were found. Actual shipping time ranges from 0 to 6 days, while scheduled shipping time ranges from 0 to 4 days. The values were retained as they are suitable for further delivery-performance analysis.

```python
df['Shipping delay days'] = (df['Days for shipping (real)'] - df['Days for shipment (scheduled)'])
```

```python
df['Shipping delay days'].value_counts().sort_index()
```

### Shipping Delay Calculation

A new `shipping_delay_days` column was created by comparing actual shipping days with scheduled shipping days.

- Negative values indicate shipments completed earlier than scheduled.
- Zero indicates shipments completed on schedule.
- Positive values indicate shipments that took longer than scheduled.

The dataset contains 103,400 records with a positive shipping delay, indicating that actual shipping time exceeded the scheduled shipping time.

```python
df['Late_delivery_risk'].value_counts()
```

```python
pd.crosstab(
    df['Late_delivery_risk'],
    df['Shipping delay days'] > 0
)
```

### Shipping Delay and Late Delivery Risk Validation

The existing `Late Delivery Risk` field was compared with the newly calculated `shipping_delay_days` metric.

The two fields are related but are not identical. All records classified with `Late Delivery Risk = 1` had a positive shipping delay. However, 4,423 records had a positive shipping delay while their `Late Delivery Risk` value was 0.

Therefore, both fields were retained:
- `Late Delivery Risk` represents the dataset's existing late-delivery classification.
- `shipping_delay_days` measures the actual difference between shipping time and scheduled shipping time.
